```python
#Importing
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv', header = 1)
```

```python
dataset.head() #To see the first few rows of the dataset.

#We will create a regression model which will predict the "temperature", so output feature is "temperature".
```

```python
dataset.info() #To know the data type of the columns
```

```python
# Now data cleaning
# This line finds all rows in the dataset that have any missing values (NaN)
dataset[dataset.isnull().any(axis=1)]
```

```python
# Set the "Region" column values for the first 123 rows (index 0 to 122) to 0
dataset.loc[:122, "Region"] = 0

# Set the "Region" column values from row 122 onwards to 1
dataset.loc[122:, "Region"] = 1

# Create a new variable 'df' as a copy of 'dataset'
df = dataset
```

```python
df.info() #Checking the data type of the "Region" column.
```

```python
df.head() #Seeing the new column "Region"
```

```python
df[['Region']] = df[['Region']].astype(int) #Changing the data type of "Region" column from float to integer
```

```python
df.info() #Checking the data type of the "Region" column.
```

```python
#Checking the null value
df.isnull().sum()
```

```python
# Removing all rows that have any missing values (NaN) from the dataframe
# Then reset the row indices so they start from 0 again
df = df.dropna().reset_index(drop=True)
```

```python
df.head() #again checking the first 5 rows
```

```python
df.isnull().sum() #Checking again for the null values, clearly no null value exists now.
```

```python
df.iloc[[122]]
#Clearly, the 122nd location is having the values as the column name, which is obviously not a good thing.
```

```python
#Dropping the 122nd row
df = df.drop(122).reset_index(drop=True)
```

```python
#Again rechecking the 122nd
df.iloc[[122]]
#now it is fine
```

```python
df.columns
#Checking the column name
#Clearly, we can see the spaces in the name of the columns, which we want to remove
```

```python
#Fixing the spaces in column names
df.columns = df.columns.str.strip()
df.columns
#now, spaces are removed.
```

```python
df.info()

#Seeing the data type, we need to convert them to integer type
```

```python
df[['month','day','year','Temperature','RH','Ws']] = df[['month','day','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
#clearly, they are now all integer (selected one)
```

```python
df.head()
#Clearly, some values are in floating too, but there data type in object
```

```python
#changing the other columns to float data type
objects = [features for features in df.columns if df[features].dtype == 'O']
```

```python
objects
#Since keep in mind that 'classes' is categorical column, so need to be followed with encoding
```

```python
for i in objects:
  if i!='Classes':
    df[i] = df[i].astype(float)
#Loop to convert the data type to float for all the columns stored in 'objects' above.
```

```python
df.describe()
```

```python
df.head()
```

```python
#Saving the cleaned data set
df.to_csv('Algerian_forest_fires_cleaned_dataset.csv', index=False)
```

```python
# Use a valid seaborn-like matplotlib style
plt.style.use('seaborn-v0_8')

# Plot histograms for all numerical columns
df_copy.hist(bins=50, figsize=(20, 15))

# Show the plots
plt.show()
```

```python
# Calculate the percentage of each class for the pie chart

# value_counts(normalize=True) gives proportions instead of counts
# Multiply by 100 to convert proportions into percentages
percentage = df_copy['Classes'].value_counts(normalize=True) * 100
```

```python
# Plotting a pie chart for the class distribution

# Labels for the pie chart categories
classlabels = ["Fire", "Not Fire"]

# Set the size of the figure
plt.figure(figsize=(12, 7))

# Create the pie chart
# percentage contains the class percentages
# labels assigns names to each slice
# autopct shows the percentage value on each slice
plt.pie(percentage, labels=classlabels, autopct="%1.2f%%")

# Add a title to the plot
plt.title("Pie Chart of Classes")

# Display the pie chart
plt.show()
```
