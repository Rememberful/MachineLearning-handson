# Label Encoding and Ordinal Encoding

Although **Label Encoding** and **Ordinal Encoding** may appear similar because both replace categories with numbers, they are **not the same** and are used in different situations.

---

# 1. What is Label Encoding?

**Label Encoding** converts each unique category into a unique integer.

Example:

| Color | Label Encoded |
| ----- | ------------: |
| Red   |             0 |
| Blue  |             1 |
| Green |             2 |

Each category gets a unique integer label.

> **Important:** The assigned numbers are identifiers, not rankings. However, many machine learning algorithms may incorrectly interpret them as having an order.

---

# 2. What is Ordinal Encoding?

**Ordinal Encoding** converts categories into numbers **while preserving their natural order or ranking**.

Example:

| Size   | Ordinal Encoded |
| ------ | --------------: |
| Small  |               1 |
| Medium |               2 |
| Large  |               3 |

Here:

Small < Medium < Large

The numerical order represents the actual relationship between the categories.

---

# 3. What Problem Do They Solve?

Machine learning algorithms cannot process text directly.

Example:

```text
Education

High School

Bachelor

Master

PhD
```

The model cannot understand these text values.

Encoding converts them into numerical values.

---

# 4. When Should I Use Them?

## Label Encoding

Use when:

* Encoding the **target variable (y)** in classification problems.
* The categories are labels where no mathematical operations are performed on the encoded values.

Example:

Spam Detection

```text
Spam

Not Spam
```

↓

```text
Spam      → 1

Not Spam  → 0
```

---

## Ordinal Encoding

Use when the feature has a **natural order**.

Examples:

Education

```text
High School

Bachelor

Master

PhD
```

Customer Rating

```text
Poor

Average

Good

Excellent
```

Satisfaction

```text
Low

Medium

High
```

---

# 5. When Should I NOT Use Them?

## Do NOT use Label Encoding

For **nominal data**.

Example:

```text
Red

Blue

Green
```

If encoded as:

```text
Red   → 0

Blue  → 1

Green → 2
```

The model may incorrectly assume:

```text
Green > Blue > Red
```

But colors have no order.

Correct method:

✅ One-Hot Encoding

---

## Do NOT use Ordinal Encoding

When categories have no natural ranking.

Example:

City

Delhi

Mumbai

Kolkata

No city is "greater" than another.

---

# 6. Types of Categorical Variables

## A. Nominal

No order.

Examples:

* Gender
* Country
* Blood Group
* Car Brand
* Color

Encoding:

✅ One-Hot Encoding

---

## B. Ordinal

Has order.

Examples:

* Education
* Shirt Size
* Customer Satisfaction
* Product Rating

Encoding:

✅ Ordinal Encoding

---

# 7. Deep Dive into Label Encoding

## Example

Animals

```text
Dog

Cat

Horse
```

Label Encoding:

| Animal | Label |
| ------ | ----: |
| Dog    |     0 |
| Cat    |     1 |
| Horse  |     2 |

The numbers are simply identifiers.

However, some algorithms may interpret:

```text
Horse > Cat > Dog
```

which is meaningless.

---

## Mathematical View

Suppose

```text
A

B

C
```

Encoded as

```text
0

1

2
```

The algorithm might infer:

```text
2 > 1 > 0
```

creating an **artificial order**.

---

# 8. Deep Dive into Ordinal Encoding

Suppose

```text
Small

Medium

Large
```

Encoding

| Size   | Value |
| ------ | ----: |
| Small  |     1 |
| Medium |     2 |
| Large  |     3 |

Here,

```text
Large > Medium > Small
```

This reflects the real-world ordering, so the numerical relationship is meaningful.

---

# 9. Label Encoding vs Ordinal Encoding

| Feature                      | Label Encoding                         | Ordinal Encoding            |
| ---------------------------- | -------------------------------------- | --------------------------- |
| Preserves Order              | ❌ No                                   | ✅ Yes                       |
| Used for Nominal Data        | ⚠️ Generally no (except target labels) | ❌ No                        |
| Used for Ordinal Data        | ❌ No                                   | ✅ Yes                       |
| Suitable for Target Variable | ✅ Yes                                  | ❌ Usually no                |
| Artificial Ordering Risk     | ✅ Yes                                  | ❌ No (order is intentional) |

# 11. Python Code

## Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Target"] = le.fit_transform(df["Target"])
```

Example:

| Target | Encoded |
| ------ | ------: |
| Spam   |       1 |
| Ham    |       0 |

---

## Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(
    categories=[
        ["Small", "Medium", "Large"]
    ]
)

df["Size"] = oe.fit_transform(df[["Size"]])
```

Output

| Size   | Encoded |
| ------ | ------: |
| Small  |       0 |
| Medium |       1 |
| Large  |       2 |

> You can specify the order explicitly through the `categories` parameter to ensure the encoder uses the intended ranking.

# 12. Pros & Cons

## Label Encoding

### Pros

* Very simple.
* Fast.
* Good for target labels.
* Memory efficient.

### Cons

* Creates artificial ordering for nominal features.
* Can mislead many algorithms if used on feature columns without order.

---

## Ordinal Encoding

### Pros

* Preserves meaningful order.
* Memory efficient.
* Suitable for ordered categories.

### Cons

* Only applicable when a true order exists.
* An incorrect ordering can hurt model performance.

---

# 15. Common Mistakes

❌ Using Label Encoding for nominal feature columns.

❌ Applying Ordinal Encoding to categories without a natural order.

❌ Forgetting to define the correct order in `OrdinalEncoder`.

❌ Assuming Label Encoding and Ordinal Encoding are interchangeable.

❌ Fitting the encoder before train-test splitting.

````python
# Import the pandas library and give it the short alias 'pd'
# pandas is used for working with structured data such as tables (DataFrames).
# 'as pd' is a convention that makes the library easier to use throughout the code.
import pandas as pd

# Create a pandas DataFrame and store it in the variable 'df'
# A DataFrame is a 2-dimensional table with rows and columns, similar to an Excel sheet or SQL table.
df = pd.DataFrame({

    # 'color' is the name of the column in the DataFrame.
    # The value associated with this key must be a list (or another iterable)
    # because each item in the list becomes one row in the column.
    'color': [

        # First row: value is 'red'
        # This is a categorical (string) value.
        'red',

        # Second row: value is 'blue'
        'blue',

        # Third row: value is 'green'
        'green',

        # Fourth row: another occurrence of 'green'
        # Duplicate values are perfectly valid in a DataFrame.
        'green',

        # Fifth row: another occurrence of 'red'
        'red',

        # Sixth row: another occurrence of 'blue'
        'blue'
    ]
})

# The resulting DataFrame looks like this:
#
#    color
# 0    red
# 1   blue
# 2  green
# 3  green
# 4    red
# 5   blue
#
# This dataset contains one categorical feature ('color').
# Since most Machine Learning algorithms cannot work directly with text,
# this column will later be converted into numerical values using Label Encoding.
````

````python
df.head()
````

````python
# Import the LabelEncoder class from sklearn.preprocessing
#
# sklearn (Scikit-learn) is one of the most popular Machine Learning libraries in Python.
# preprocessing is a module that contains tools for preparing data before training a model.
# LabelEncoder is a class used to convert categorical (text) labels into numerical labels.
#
# Example:
# 'blue'  -> 0
# 'green' -> 1
# 'red'   -> 2
#
# The assigned numbers are based on the alphabetical order of the unique values,
# not on the order in which they appear in the dataset.
from sklearn.preprocessing import LabelEncoder

# Create an object (instance) of the LabelEncoder class and store it
# in the variable 'lbl_encoder'.
#
# LabelEncoder() has no parameters here because the default behavior
# is sufficient for basic label encoding.
#
# This object will learn the unique categories from the data when
# we later call the .fit() or .fit_transform() method.
#
# We create this object once so that it can be reused for encoding data.
lbl_encoder = LabelEncoder()
````

````python
# Apply Label Encoding to the 'color' column.
#
# fit_transform() performs TWO operations in one step:
#
# 1. fit()
#    - Reads the data.
#    - Finds all unique categories.
#    - Sorts them alphabetically.
#    - Assigns an integer to each category.
#
#      For this dataset:
#      'blue'  -> 0
#      'green' -> 1
#      'red'   -> 2
#
# 2. transform()
#    - Replaces every category with its assigned integer.
#
# IMPORTANT:
# LabelEncoder is designed to work with a **1-dimensional array (Series)**,
# but df[['color']] returns a **2-dimensional DataFrame** with shape (6, 1).
# Therefore, this line will raise a ValueError.
#
# Correct usage:
# lbl_encoder.fit_transform(df['color'])
#
# Here:
# - fit_transform(data)
#     data : A 1-dimensional array-like object containing categorical values
#            to be learned and converted into numbers.
#
# The expected output (using df['color']) would be:
# array([2, 0, 1, 1, 2, 0])
lbl_encoder.fit_transform(df[['color']])
````

````python
# Use the already trained LabelEncoder to convert a new category ('red')
# into its corresponding numerical label.
#
# transform() DOES NOT learn anything new.
# It only uses the mapping that was created earlier by fit() or fit_transform().
#
# Since the encoder has already learned:
# 'blue'  -> 0
# 'green' -> 1
# 'red'   -> 2
#
# It will return:
# array([2])
#
# IMPORTANT:
# LabelEncoder expects a **1-dimensional array-like object** as input.
# Here, [['red']] is a **2-dimensional list**, which is NOT the expected format
# and will raise a ValueError.
#
# Correct usage:
# lbl_encoder.transform(['red'])
#
# Parameter:
# - ['red']
#     A 1-dimensional list containing the category (or categories) to encode.
#     You can also pass multiple values, for example:
#     lbl_encoder.transform(['red', 'blue', 'green'])
#
# Note:
# If you try to transform a category that the encoder has never seen during
# fit() (for example, 'yellow'), LabelEncoder will raise a ValueError because
# it cannot assign a label to an unknown category.
lbl_encoder.transform([['red']])
````

````python
# --------------------------- Ordinal Encoding ---------------------------
# Ordinal Encoding is used when the categories have a meaningful order.
#
# Example:
# small < medium < large
#
# Unlike LabelEncoder (which assigns numbers alphabetically),
# OrdinalEncoder allows us to preserve the actual order of the categories.

# Import the pandas library and give it the alias 'pd'.
# pandas is used for creating and manipulating tabular data.
import pandas as pd

# Import the OrdinalEncoder class from sklearn.preprocessing.
#
# sklearn (Scikit-learn) is a Machine Learning library.
# preprocessing contains tools used to prepare data before training a model.
#
# OrdinalEncoder converts ordered categorical values into numerical values
# while preserving their order.
from sklearn.preprocessing import OrdinalEncoder

# Create a DataFrame and store it in the variable 'df'.
# A DataFrame is a 2-dimensional table consisting of rows and columns.
df = pd.DataFrame({

    # 'size' is the name of the column.
    # Each element in the list becomes one row in the DataFrame.
    'size': [

        # First row
        'small',

        # Second row
        'medium',

        # Third row
        'large',

        # Fourth row
        'medium',

        # Fifth row
        'small',

        # Sixth row
        'large'
    ]
})

# The DataFrame looks like:
#
#      size
# 0   small
# 1  medium
# 2   large
# 3  medium
# 4   small
# 5   large
#
# Since the categories have a natural order:
#
# small < medium < large
#
# this is an example of an ORDINAL categorical feature,
# making it suitable for Ordinal Encoding.
````

````python
df
````

````python
# assign rankings

# Create an instance and then fit_transform
````

````python
# Create an object (instance) of the OrdinalEncoder class
# and store it in the variable 'encoder'.
#
# We create this object so that it can learn the category order
# and later transform categorical values into numerical values.
encoder = OrdinalEncoder(

    # categories is a parameter used to specify the exact order
    # of the categories for each feature (column).
    #
    # Syntax:
    # categories = [list_for_column1, list_for_column2, ...]
    #
    # Since our DataFrame has only ONE column ('size'),
    # we provide only ONE list inside another list.
    #
    # Outer list  -> represents all columns.
    # Inner list  -> represents the ordered categories for the 'size' column.
    #
    # If there were two columns, for example:
    # categories=[
    #     ['small', 'medium', 'large'],
    #     ['low', 'average', 'high']
    # ]
    #
    # then the first list would belong to the first column,
    # and the second list would belong to the second column.
    categories=[

        # This list defines the desired order of the categories.
        #
        # The encoder will assign:
        # 'small'  -> 0
        # 'medium' -> 1
        # 'large'  -> 2
        #
        # These numbers are assigned according to the order YOU specify,
        # not alphabetical order.
        ['small', 'medium', 'large']
    ]
)

# No encoding happens at this stage.
#
# This line only creates the encoder object and stores the category order.
# The actual encoding will happen later when we call:
#
# encoder.fit()
# or
# encoder.fit_transform()
#
# Using the custom category order ensures that the encoded numbers
# correctly represent the real ordinal relationship:
#
# 0 < 1 < 2
# ↑   ↑   ↑
# small  medium  large
````

````python
# Apply Ordinal Encoding to the 'size' column.
#
# fit_transform() performs TWO operations in one step:
#
# 1. fit()
#    - Reads the data from the 'size' column.
#    - Learns the categories based on the order specified earlier:
#          'small'  -> 0
#          'medium' -> 1
#          'large'  -> 2
#
#    - Unlike LabelEncoder, OrdinalEncoder does NOT sort categories
#      alphabetically when you explicitly provide the 'categories' parameter.
#      It strictly follows the order you specified.
#
# 2. transform()
#    - Replaces each categorical value with its corresponding numerical value.
#
# Parameter:
# df[['size']]
# - df            : The DataFrame containing the data.
# - [['size']]    : Selects the 'size' column as a DataFrame (2-dimensional).
#
# Why use double square brackets [[]] instead of single []?
#
# df['size']      -> Returns a pandas Series (1-dimensional).
# df[['size']]    -> Returns a pandas DataFrame (2-dimensional).
#
# OrdinalEncoder expects a 2-dimensional input because it is designed
# to encode one or more feature columns at the same time.
# Therefore, using df[['size']] is the correct approach.
#
# Expected output:
#
# array([
#     [0.],
#     [1.],
#     [2.],
#     [1.],
#     [0.],
#     [2.]
# ])
#
# Notice:
# - The output contains decimal numbers (0.0, 1.0, 2.0).
# - This is because OrdinalEncoder returns a NumPy array of type float by default.
#
# Final mapping:
# small  -> 0.0
# medium -> 1.0
# large  -> 2.0
encoder.fit_transform(df[['size']])
````

````python
# what is the value assigned to the small?

encoder.transform([['small']])

# 0
````

| Property             | Label Encoding     | Ordinal Encoding               |
| -------------------- | ------------------ | ------------------------------ |
| Mathematical mapping | Category → integer | Category → ordered integer     |
| Formula              | (f(x)=n)           | (f(x_i)<f(x_j)) preserves rank |
| Order exists?        | ❌ No               | ✅ Yes                          |
| Numbers represent    | IDs                | Rankings                       |
| Example              | Red=0, Blue=1      | Low=0, Medium=1, High=2        |
| Used for             | Target classes     | Ordered features               |
